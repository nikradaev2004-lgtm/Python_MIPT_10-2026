```python
def pascal_tringle(row):
  l_total=[]
  l_1 = [1]
  l_total.append(l_1)
  if row == 1:
    return l_total
  l = l_1.copy()
  l.append(1)
  l_total.append(l)
  if row == 2:
    return l_total
  if row > 2:
    #l = [1, 1]
    for a in range (2, row):
      l_new = []
      l_new.append(1)
      for i in range(0, len(l)-1):
        l_new.append(l[i]+l[i+1])
      l_new.append(1)
      l_total.append(l_new)
      l=l_new
  return l_total
```
