```python
"""1. Celsius and Fahrenheit Converter

Description: To convert temperature between Fahrenheit to Celsius and Celsius to Fahrenheit
"""
while True:
  a = input("Type F to convert Celsius to Fahrenheit and C to convert Fahrenheit to Celsius: ")
  if a=="F" or a=="f": # f for user comfort
    start_temperature = float(input("Enter temperature in Celsius: "))
    print(f"Temperature in Fahrenheit: {round(start_temperature*1.8+32, 2)}")
  if a=="C" or a=="c": # c for user comfort
    start_temperature = float(input("Enter temperature in Fahrenheit: "))
    print(f"Temperature in Celsius: {round((start_temperature-32)*5/9, 2)}")
  print()
```
