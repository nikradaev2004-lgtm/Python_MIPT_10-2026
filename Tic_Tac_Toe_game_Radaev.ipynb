```python
def winner(moves: list[list[int]]):
  A_moves = moves[::2] # разделение на А и B игроков
  B_moves = moves[1::2]

  first_elements_A = [sub[0] for sub in A_moves] # проверка на горизонтальные и вертикальные линии. Для этого все первые или все вторые числа в строке должны быть равны
  second_elements_A = [sub[1] for sub in A_moves]
  counts_A = []
  counts_A.append(first_elements_A.count(0))
  counts_A.append(first_elements_A.count(1))
  counts_A.append(first_elements_A.count(2))
  counts_A.append(second_elements_A.count(0))
  counts_A.append(second_elements_A.count(1))
  counts_A.append(second_elements_A.count(2))
  if 3 in counts_A:
    return "A"

  first_elements_B = [sub[0] for sub in B_moves] # проверка на горизонтальные и вертикальные линии. Для этого все первые или все вторые числа в строке должны быть равны
  second_elements_B = [sub[1] for sub in B_moves]
  counts_B = []
  counts_B.append(first_elements_B.count(0))
  counts_B.append(first_elements_B.count(1))
  counts_B.append(first_elements_B.count(2))
  counts_B.append(second_elements_B.count(0))
  counts_B.append(second_elements_B.count(1))
  counts_B.append(second_elements_B.count(2))
  if 3 in counts_B:
    return "B"


  if ([0,0] and [1,1] and [2,2]) in A_moves: #проверка на косые линии
    return "A"
  if ([2,0] and [1,1] and [0,2]) in A_moves: #проверка на косые линии
    return "A"

  if ([0,0] and [1,1] and [2,2]) in B_moves: #проверка на косые линии
    return "B"
  if ([2,0] and [1,1] and [0,2]) in B_moves: #проверка на косые линии
    return "B"

  else:
    return "Draw"
```
