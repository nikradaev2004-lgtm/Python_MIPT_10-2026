```python
"""
ПРАВИЛА ДЕРЕВА
Каждый узел окрашен либо в красный, либо в чёрный цвет.
Корень дерева всегда чёрный.
Все листья NIL (отсутствующие узлы) считаются чёрными.
У красного узла оба потомка должны быть чёрными.
На любом пути от узла до NIL-листьев количество чёрных узлов одинаково.
"""
```

```python
class RedBlackTree:
  def __init__(self):
    self.root = None
```

```python
class Node:
  def __init__(self, number, color="red"):
        self.number = number
        self.color = color
        self.left_child = None
        self.right_child = None
        self.parent = None
```

```python
"""
Для балансировки часто используются повороты
Левый поворот: нижний элемент верх, верхний элемент налево (местами обмениваются родители и дети)
Правый поворот (зеркальный): верхний элемент вниз, нижний элемент направо (местами обмениваются родители и дети)
"""
```

```python
def right(self, Node):
    parent = Node.parent

    if parent.parent is None:
        self.root = Node # поднявшись вверх, становится корнем
        Node.parent = None # поднявшись вверх, становится корнем (следовательно, родителя)
    else:
        Node.parent = parent.parent

        if parent == parent.parent.left_child: #добавляем деда и его переход
            parent.parent.left_child = Node
        else:
            parent.parent.right_child = Node

    parent.left_child = Node.right_child

    if Node.right_child: # если у основного узла есть правый ребенок
        Node.right_child.parent = parent

    Node.right_child = parent # итоговая картина
    parent.parent = Node # итоговая картина
```

```python
def left(self, Node):
    parent = Node.parent

    if parent.parent is None:
        self.root = Node # поднявшись вверх, становится корнем
        Node.parent = None # у нового корня нет родителя
    else:
        Node.parent = parent.parent

        if parent == parent.parent.left_child: # добавляем деда и его переход
            parent.parent.left_child = Node
        else:
            parent.parent.right_child = Node

    parent.right_child = Node.left_child

    if Node.left_child: # если у основного узла есть левый ребенок
        Node.left_child.parent = parent

    Node.left_child = parent # итоговая картина
    parent.parent = Node # итоговая картина
```

```python
def insert(self, value):
    new_node = Node(value)
    current = self.root
    parent = None

    while current is not None:
        parent = current

        if value < current.number:
            current = current.left_child

        elif value > current.number:
            current = current.right_child

        else:
            return

    new_node.parent = parent

    if parent is None:
        self.root = new_node
    elif value < parent.number:
        parent.left_child = new_node
    else:
        parent.right_child = new_node

    self.rearangement_first_type(new_node) # балансировка после вставки
```

```python
# Красный дядя
"""
Родителя (P) перекрашиваем в чёрный
Дядю (U) перекрашиваем в чёрный
Дедушку (G) перекрашиваем в красный
Продолжаем проверку от дедушки (G)
!!!!! НЕТ РАЗНИЦЫ - ДОБАВИЛСЯ НОД СЛЕВА ИЛИ СПРАВА
"""
```

```python
def rearangement_first_type(self, Node): # red unckle
  while Node != self.root and Node.parent.color == "red":
      parent = Node.parent
      grandparent = parent.parent

      if parent == grandparent.left_child:
        uncle = grandparent.right_child
      else:
        uncle = grandparent.left_child

      if uncle is not None and uncle.color == "red":
        parent.color = "black"
        uncle.color = "black"
        grandparent.color = "red"
        Node = grandparent
      else:
        self.rearangement_third_type(Node) # дядя чёрный
        break

  self.root.color = "black" # корень всегда черный!!!
```

```python
# Дядя чёрный, образуется линия. Красный нод и красный родитель
"""
Родителя (P) перекрашиваем в чёрный
Дедушку (G) перекрашиваем в красный
Выполняем правый поворот относительно дедушки (G)
"""
```

```python
def rearangement_second_type(self, Node): # дядя черный, линия
  while Node != self.root and Node.parent.color == "red":
      parent = Node.parent
      grandparent = parent.parent

      parent.color = "black"
      grandparent.color = "red"

      if parent == grandparent.left_child:
        self.right(parent) # функция разворота направо
      else:
        self.left(parent) # функция разворота налево

  self.root.color = "black" # корень всегда черный!!!
```
