```python
from datetime import date, timedelta

def busiest_day(visits):
  counts = []
  earliest_day = min(date.fromisoformat(arrival) for arrival, _ in visits) # поиск самого раннего дня в кортеже
  last_day = max(date.fromisoformat(departure) for _, departure in visits) # поиск самого позднего дня в кортеже


  #Цикл:
  day = earliest_day
  while day <= last_day: # пока счетчик самого раннего дня в кортеже не равен последнему дню в кортеже. Шаг +1 (один день)
      score = 0 # каждый день счетчик обновляется
      for arrival, departure in visits:
          if date.fromisoformat(arrival) <= day <= date.fromisoformat(departure):
              score = score + 1 # каждый день происходит подсчет, сколько всего гостей в отеле
      counts.append(score)
      day = day + timedelta(days=1) # переход на следующий день в цикле. Шаг +1 (один день)


  # Определим день с макимальным количеством посетителей
  if counts.count(max(counts)) == 1: # counts - список score посетителей за каждый день
    days_past = counts.index(max(counts))
    answer_day = earliest_day + timedelta(days=days_past)
    return answer_day # the busiest_day (the answer)
  else: # несколько дней, в которых было максимальное количество гостей
    list_of_answers = []
    for i in range(0, len(counts)):
      if counts[i] == max(counts):
        answer_day = earliest_day + timedelta(days=i)
        list_of_answers.append(answer_day)
    return list_of_answers # the busiest days (the answers)
```

```python
visits = [("2024-09-15", "2024-09-15"), ("2024-09-14", "2024-09-21")]
print(busiest_day(visits))
```
