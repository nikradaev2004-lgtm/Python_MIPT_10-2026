```python
# Динамическое программирование

def max_profit(prices, k): # сообщается список цен (цена меняется каждый день; один день - одна цена)
    cash = [0] * (k + 1) # это наши полученные средства (выручка). Изначально - линия из нулей, так как мы изначально ничего не заработали
    # Элементов k, так как это максимальное число операций, которые мы можем сделать (купить и продать акции)

    hold = [float('-inf')] * (k + 1) # это наши потраченные на покупку акций средства. Изначально - линия из минус бескончностей, чтобы они на "победили"
    # при использовании функции max
    # Элементов k, так как это максимальное число операций, которые мы можем сделать (купить и продать акции)

    for i in range(len(prices)): # прогоняется за каждый день
        for j in range(k, 0, -1): # програняется для каждого элемента вазможного решения (возможные решения о покупке или продаже). Остается только max (максимум)
            cash[j] = max(cash[j], hold[j] + prices[i])
            hold[j] = max(hold[j], cash[j - 1] - prices[i])
    return max(cash) # ответ - максимальная выручка
```

```python
# Пример проверки функции динамического программирования
prices = [3, 2, 6, 5, 0, 3]
k = 2

print(max_profit(prices, k))
```
