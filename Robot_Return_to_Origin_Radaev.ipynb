```python
def robot(text):
  x = 0
  y = 0
  for command in text:
    if command == "R":
      x=x+1
    if command == "L":
      x=x-1
    if command == "U":
      y=y+1
    if command == "D":
      y=y-1
  if x==0 and y==0:
    return True
  else:
    return False

```

```python
robot("LL")
```
