```python
# Radaev Nikita, 3-XI-2026
import heapq

n = int(input()) # количество объектов в а, которые будут учитываться для коэффициентов значимости признаков
if n>=1 and n<=100:
  a = list(map(int, input().split()))
  if len(a)==n and all(0 <= x <= 100_000_000 for x in a):
    d = int(input()) # количество объектов для ранжирования
    features = []
    for i in range (0, d):
      feature_unit = list(map(int, input().split()))
      if len(feature_unit) == n: # или == len(a), или len(features) == n
        features.append(feature_unit)
    # создали список значений features. Рассчитавем значимость каждого элемента
    scores = []
    for i in range(d):
      score = sum(a[j] * features[i][j] for j in range(n))
      scores.append(score)
    # получен ранжированный список scores
# перенесем список в кучу (heap)
    heap = []
    for i in range(d):
      heapq.heappush(heap, (-scores[i], i)) # добавляем в кучу отрицательные значения, чтобы наибольшие элементы были в вершине дерева
      # Использование отрицательных чисел необходимо для использование стандартного синтаксиса модуля heapq (import heapq)
    q = int(input()) # количество запросов к ранжированию
    if q>=1 and q<=100_000:
      for i in range (q):
        query = list(map(int, input().split()))
        if len(query) == 2:
          k = query[1]
          print(*(x[1] + 1 for x in heapq.nsmallest(k, heap))) # берём k наименьших (т.е. с наибольшим score) элементов
        if len(query) == 4:
          i_second_option = query[1]
          j_second_option = query[2]
          v_second_option = query[3]
        # извлечение и удаление из кучи нужного объекта
          for x in range(len(heap)):
            if heap[x][1] == i_second_option - 1:
              heap.pop(x) # удаление элемента, который нужно изменить
              break
          heapq.heapify(heap) # нормализация кучи
          number_to_change = i_second_option - 1 # выделение порядкового номера нужного числа
          root_numbers = features[number_to_change] # поиск чисел, по которым был сосчитан score нужного нам числа
          root_numbers[j_second_option - 1] = v_second_option # изменение номера j на значение v
          new_score = sum(a[j] * root_numbers[j] for j in range(n)) # подсчет нового значения score
          heapq.heappush(heap, (-new_score, number_to_change)) # добавление измененного объекта (score и порядковый номер) в кучу
          print(*(x[1] for x in sorted(heap))) # вывод порядковых номеров после изменений
```
