```python
def canPlaceFlowers(flowerbed: list[int], n: int):
  flowerbed.append(2)
  flowerbed = " ".join(map(str, flowerbed))
  flowerbed = flowerbed.replace(" ", "")
  while n>0:
    if "0000" in flowerbed:
      n=n-1
      flowerbed = flowerbed.replace("00", "", 1)
    if "000" in flowerbed:
      n=n-1
      flowerbed = flowerbed.replace("000", "", 1)
      print(flowerbed)
    else:
      return False
  return True
```
