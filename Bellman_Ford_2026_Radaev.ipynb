```python
#Как функция
def bellman_ford(n, edges, start):
    dist = [10**12] * n # расстояния до всех вершин - бесконечность (крайне большое счисло)
    dist[start] = 0 # до самой себя расстояние 0
    parent = [-1] * n # из какой вершины пришли (-1 = неизвестно)

    for i in range(n - 1):
        for u, v, w in edges: # в отличие от алгоритма Дейкстра работает вслепую много раз
            if dist[u] != 10**12 and dist[u] + w < dist[v]: # обращение по номерам элементов. dist[u] - номер начала пути графа
                dist[v] = dist[u] + w # изменение на лучшее расстояние
                parent[v] = u # сохраняются вершины, откуда пришли

    return dist, parent
```

```python
n = 5
edges = [
    (0, 1, 6), (0, 2, 7),
    (1, 2, 8), (1, 3, 5), (1, 4, -4),
    (2, 3, -3), (2, 4, 9),
    (3, 1, -2),
    (4, 3, 7),
]
start = 0

dist = [10**12] * n # расстояния до всех вершин - бесконечность (крайне большое счисло)
dist[start] = 0 # до самой себя расстояние 0
parent = [-1] * n # из какой вершины пришли (-1 = неизвестно)

for i in range(n - 1):
    for u, v, w in edges: # в отличие от алгоритма Дейкстра работает вслепую много раз
        if dist[u] != 10**12 and dist[u] + w < dist[v]: # обращение по номерам элементов. dist[u] - номер начала пути графа
            dist[v] = dist[u] + w # изменение на лучшее расстояние
            parent[v] = u # сохраняются вершины, откуда пришли

print(dist)
```

```python
def get_path(parent, dist, start, end):
    # если вершина недостижима, пути нет
    if dist[end] == 10**12:
        return []

    path = []
    x = end # end - вершина, до которой нужно восстановить путь
    while x != -1: # идём назад по parent до старта
        path.append(x)
        x = parent[x]

    path.reverse() # разворачиваем: от start к end
    return path
```

```python
for v in range(n):
    path = get_path(parent, dist, start, v)
    if path == []:
        print("Вершина", v, ": недостижима")
    else:
        print("Вершина", v, ": расстояние =", dist[v], ", путь =", path)
```
